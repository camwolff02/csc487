## Adding Classical Features to the Resnet50 Classifier

```python
class ExtendedImageDataset(torch.utils.data.Dataset):
    def __init__(self, df_labels, img_dir, transform=None, target_transform=None):
        self.img_names = df_labels["image"]
        self.img_labels = df_labels.drop(pl.col("image")).to_torch().to(device)
        self.img_dir = img_dir
        self.transform = transform
        self.target_transform = target_transform

    def __len__(self):
        return len(self.img_labels)

    def __getitem__(self, idx):
        img_path = os.path.join(self.img_dir, self.img_names[idx] + ".jpg")
        image = torchvision.io.read_image(img_path)
        label = self.img_labels[idx]

        # Calculating additional features
        binary_mask = segment_image(image)
        vector = torch.tensor(describe_skin_lesion(image, binary_mask)).to(device)
        
        if self.transform:
            image = self.transform(image)
        if self.target_transform:
            label = self.target_transform(label)
        return (image, vector), label
```

```python
class ExtendedResnet(torch.nn.Module):
    def __init__(self):
        super(ExtendedResnet, self).__init__()
        self.device = device
        self.resnet = torchvision.models.resnet50(weights=torchvision.models.ResNet50_Weights.IMAGENET1K_V2)
        num_features = resnet.fc.in_features
        self.resnet.fc = nn.Linear(num_features, 8)

        self.mlp = nn.Sequential(
            nn.Linear(7, 8),
            nn.ReLU(),
        )

        self.fc = nn.Sequential(
            nn.Linear(16, 8),
            nn.ReLU(),
            nn.Linear(8, 8),
        )

    def forward(self, x):
        image, vec = x
        image_pred = self.resnet(image)
        vec_pred = self.mlp(vec)
        return self.fc(image_pred + vec_pred)
```

```python
import os
from ultralytics import YOLO
```

```python
model = YOLO('yolo11n.pt')
model.train(
    data=f'{os.getcwd()}/combined_dataset/data.yaml', 
    epochs=400,  # seems to converge at 397
    batch=48,  # Max I can do with my VRAM and still use my PC
    save_period=50,
    device='cuda',
    multi_scale=True,  # randomly scale images up and down during training
    plots=True,
)
```

```python
model.export(format='onnx')
```
